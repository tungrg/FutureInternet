## Exercise 1.1: Implementation of Random Access and Slotted ALOHA
- Simulate $100 - 1000$ messages per hour ($T = 3600\text{ s}$) in steps of $100$ messages
- Fixed transmission time $T_s = 0.999\text{ s}$, slot length $= 1.0\text{ s}$
- Compare simulated collision probability with the analytical formula:
  $$p = 1 - \prod_{a \in S \setminus \{s\}} \left(1 - \frac{T_s + T_a}{T}\right) = 1 - \left(1 - \frac{2 T_s}{T}\right)^{N-1}$$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from aloha_simulation import run_sweep
```

```python
num_messages_list = np.arange(100, 1001, 100)
simulation_time = 3600
transmission_time = 0.999
time_slot = 1.0
num_trials = 100

result_pure_1_1, result_slotted_1_1 = run_sweep(
    message_counts=num_messages_list,
    simulation_time=simulation_time,
    transmission_time=transmission_time,
    time_slot=time_slot,
    trials=num_trials,
    seed=42,
)

df_pure_1_1 = pd.DataFrame(result_pure_1_1)
df_slotted_1_1 = pd.DataFrame(result_slotted_1_1)
```

```python
for protocol, results in (("Pure ALOHA", result_pure_1_1), ("Slotted ALOHA", result_slotted_1_1)):
    print(f"\n{protocol} (averaged over {num_trials} trials; no retransmissions)")
    for result in results:
        print(
            f"N={result['messages']:4d}, G={result['offered_load']:.3f}, "
            f"mean collided={result['mean_collisions']:5.1f} (±{result['std_collisions']:4.1f}), "
            f"collision prob={result['collision_probability']:.4f} "
            f"(theory={result['theoretical_collision_probability']:.4f}), "
            f"utilization={100 * result['channel_utilization']:.2f}%"
        )
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Plot 1: Number of Collisions
axes[0].errorbar(
    df_pure_1_1["messages"],
    df_pure_1_1["mean_collisions"],
    yerr=df_pure_1_1["std_collisions"],
    marker="o",
    capsize=3,
    label="Pure ALOHA (sim ± 1 std)",
)
axes[0].errorbar(
    df_slotted_1_1["messages"],
    df_slotted_1_1["mean_collisions"],
    yerr=df_slotted_1_1["std_collisions"],
    marker="s",
    capsize=3,
    label="Slotted ALOHA 1s (sim ± 1 std)",
)
axes[0].set_title("Exercise 1.1: Number of Collided Messages")
axes[0].set_xlabel("Messages per hour (N)")
axes[0].set_ylabel("Collided messages")
axes[0].grid(True, alpha=0.3)
axes[0].legend()

# Plot 2: Collision Probability (Simulated vs Theoretical)
axes[1].errorbar(
    df_pure_1_1["messages"],
    df_pure_1_1["collision_probability"],
    yerr=df_pure_1_1["std_collision_probability"],
    marker="o",
    capsize=3,
    label="Pure ALOHA (simulated)",
)
axes[1].plot(
    df_pure_1_1["messages"],
    df_pure_1_1["theoretical_collision_probability"],
    "k--",
    label="Pure ALOHA (theoretical eq.)",
)
axes[1].errorbar(
    df_slotted_1_1["messages"],
    df_slotted_1_1["collision_probability"],
    yerr=df_slotted_1_1["std_collision_probability"],
    marker="s",
    capsize=3,
    label="Slotted ALOHA 1s (simulated)",
)
axes[1].plot(
    df_slotted_1_1["messages"],
    df_slotted_1_1["theoretical_collision_probability"],
    "r:",
    label="Slotted ALOHA 1s (theoretical)",
)
axes[1].set_title("Exercise 1.1: Collision Probability")
axes[1].set_xlabel("Messages per hour (N)")
axes[1].set_ylabel("Collision probability p")
axes[1].grid(True, alpha=0.3)
axes[1].legend()

plt.tight_layout()
plt.show()
```

### Key Insights (Exercise 1.1)
1. **Halved Vulnerable Period in Slotted ALOHA**: In Pure ALOHA, a frame of duration $T_s = 0.999\text{ s}$ collides with any other frame starting within $(-T_s, +T_s)$ of its start time (a vulnerable window of $2 T_s \approx 2\text{ s}$). Synchronizing transmissions into $1\text{ s}$ slots restricts collisions to frames choosing the exact same slot (vulnerable window of $1\text{ s}$).
2. **Lower Collision Probability & Higher Throughput**: Across all message counts $N \in [100, 1000]$, Slotted ALOHA exhibits roughly half the collision probability of Pure ALOHA at lower loads and significantly higher successful channel utilization.
3. **Analytical Agreement**: The simulated Pure ALOHA collision probabilities closely match the analytical equation $p = 1 - \prod_{a \in S \setminus \{s\}} \left(1 - \frac{T_s + T_a}{T}\right)$.

## Exercise 1.2: Variable Transmission Time $T_s \sim \mathcal{U}(0, 2)\text{ s}$
- Transmission time for each message is drawn uniformly at random between $0\text{ s}$ and $2\text{ s}$
- Test Slotted ALOHA with slot lengths of $1\text{ s}$ and $2\text{ s}$ and compare against Exercise 1.1

```python
variable_tx_range = (0.0, 2.0)

# Run Exercise 1.2 with slot length = 1.0s
result_pure_1_2, result_slotted_1_2_slot1 = run_sweep(
    message_counts=num_messages_list,
    simulation_time=simulation_time,
    transmission_time=variable_tx_range,
    time_slot=1.0,
    trials=num_trials,
    seed=42,
)

# Run Exercise 1.2 with slot length = 2.0s
_, result_slotted_1_2_slot2 = run_sweep(
    message_counts=num_messages_list,
    simulation_time=simulation_time,
    transmission_time=variable_tx_range,
    time_slot=2.0,
    trials=num_trials,
    seed=42,
)

df_pure_1_2 = pd.DataFrame(result_pure_1_2)
df_slotted_1_2_slot1 = pd.DataFrame(result_slotted_1_2_slot1)
df_slotted_1_2_slot2 = pd.DataFrame(result_slotted_1_2_slot2)
```

```python
for label, results in (
    ("Pure ALOHA (Tx ~ U(0, 2)s)", result_pure_1_2),
    ("Slotted ALOHA (Tx ~ U(0, 2)s, slot = 1s)", result_slotted_1_2_slot1),
    ("Slotted ALOHA (Tx ~ U(0, 2)s, slot = 2s)", result_slotted_1_2_slot2),
):
    print(f"\n{label} (averaged over {num_trials} trials)")
    for result in results:
        print(
            f"N={result['messages']:4d}, "
            f"mean collided={result['mean_collisions']:5.1f} (±{result['std_collisions']:4.1f}), "
            f"collision prob={result['collision_probability']:.4f}, "
            f"utilization={100 * result['channel_utilization']:.2f}%"
        )
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Left: Collision probability comparison between Ex 1.1 and Ex 1.2
axes[0].plot(
    df_pure_1_1["messages"],
    df_pure_1_1["collision_probability"],
    "o--",
    label="Ex 1.1: Pure ALOHA (Tx=0.999s)",
)
axes[0].plot(
    df_slotted_1_1["messages"],
    df_slotted_1_1["collision_probability"],
    "s--",
    label="Ex 1.1: Slotted ALOHA (Tx=0.999s, slot=1s)",
)
axes[0].plot(
    df_pure_1_2["messages"],
    df_pure_1_2["collision_probability"],
    "o-",
    label="Ex 1.2: Pure ALOHA (Tx~U(0,2)s)",
)
axes[0].plot(
    df_slotted_1_2_slot1["messages"],
    df_slotted_1_2_slot1["collision_probability"],
    "^-",
    label="Ex 1.2: Slotted ALOHA (Tx~U(0,2)s, slot=1s)",
)
axes[0].plot(
    df_slotted_1_2_slot2["messages"],
    df_slotted_1_2_slot2["collision_probability"],
    "d-",
    label="Ex 1.2: Slotted ALOHA (Tx~U(0,2)s, slot=2s)",
)
axes[0].set_title("Exercise 1.1 vs 1.2: Collision Probability")
axes[0].set_xlabel("Messages per hour (N)")
axes[0].set_ylabel("Collision probability p")
axes[0].grid(True, alpha=0.3)
axes[0].legend(fontsize=8)

# Right: Number of collisions in Ex 1.2
axes[1].errorbar(
    df_pure_1_2["messages"],
    df_pure_1_2["mean_collisions"],
    yerr=df_pure_1_2["std_collisions"],
    marker="o",
    capsize=3,
    label="Pure ALOHA (Tx~U(0,2)s)",
)
axes[1].errorbar(
    df_slotted_1_2_slot1["messages"],
    df_slotted_1_2_slot1["mean_collisions"],
    yerr=df_slotted_1_2_slot1["std_collisions"],
    marker="^",
    capsize=3,
    label="Slotted ALOHA (slot=1s)",
)
axes[1].errorbar(
    df_slotted_1_2_slot2["messages"],
    df_slotted_1_2_slot2["mean_collisions"],
    yerr=df_slotted_1_2_slot2["std_collisions"],
    marker="d",
    capsize=3,
    label="Slotted ALOHA (slot=2s)",
)
axes[1].set_title("Exercise 1.2: Collided Messages (with ±1 std)")
axes[1].set_xlabel("Messages per hour (N)")
axes[1].set_ylabel("Collided messages")
axes[1].grid(True, alpha=0.3)
axes[1].legend()

plt.tight_layout()
plt.show()
```

### Key Insights (Exercise 1.2 & Bonus)
1. **Pure ALOHA is Insensitive to Packet-Length Variance (at Fixed Mean)**: Because $T_s \sim \mathcal{U}(0, 2)\text{ s}$ has mean $\mathbb{E}[T_s] = 1.0\text{ s} \approx 0.999\text{ s}$, the expected vulnerable window $\mathbb{E}[T_s + T_a] = 2.0\text{ s}$ is unchanged from Exercise 1.1. Pure ALOHA's collision probability in Exercise 1.2 is virtually identical to Exercise 1.1.
2. **Slotted ALOHA with $1\text{ s}$ Slots Suffers from Slot Spillover**: Half of the messages have duration $T_s > 1\text{ s}$ and spill over into the next time slot, colliding with frames in both slot $k$ and slot $k+1$. Its collision probability jumps dramatically compared to Exercise 1.1.
3. **Slotted ALOHA with $2\text{ s}$ Slots Eliminates Spillover but Halves the Number of Slots**: Sizing the slot to the maximum frame duration ($2\text{ s}$) prevents cross-slot collisions, but reduces the number of slots per hour from $3600$ to $1800$ and wastes airtime whenever $T_s < 2\text{ s}$ (on average $50\%$ of each active slot is empty guard/padding time). Consequently, its collision probability $1 - (1 - 1/1800)^{N-1}$ matches Pure ALOHA's collision probability!
4. **Bonus (Repeating Simulations)**: A single Monte Carlo run has significant random variance. Repeating each configuration over `num_trials = 100` independent runs with a seeded RNG and reporting the **mean and standard deviation (or confidence intervals)** stabilizes the curves and allows statistically reliable comparisons.
